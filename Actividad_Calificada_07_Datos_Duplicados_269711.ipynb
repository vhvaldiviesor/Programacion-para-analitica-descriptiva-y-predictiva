## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# keep='first': marca como duplicada cada copia EXCEPTO la primera aparición de cada grupo
# -> nos dice cuántas filas "sobran" si conserváramos solo la primera ocurrencia de cada reserva repetida
dup_first = hoteles.duplicated(keep='first').sum()

# keep='last': marca como duplicada cada copia EXCEPTO la última aparición de cada grupo
# -> el mismo número de filas "sobrantes", pero contadas desde el otro extremo del grupo
dup_last = hoteles.duplicated(keep='last').sum()

# keep=False: marca como duplicada TODA fila que pertenece a un grupo con más de una copia
# (incluida la primera y la última) -> nos dice cuántas filas están *involucradas* en algún duplicado
dup_false = hoteles.duplicated(keep=False).sum()

print(f"keep='first' -> {dup_first} filas duplicadas (sobrantes si se conserva la primera copia de cada grupo)")
print(f"keep='last'  -> {dup_last} filas duplicadas (sobrantes si se conserva la última copia de cada grupo)")
print(f"keep=False   -> {dup_false} filas duplicadas (TODAS las copias de cada grupo repetido, sin conservar ninguna)")
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Porcentaje de filas sobrantes: usamos dup_first (o dup_last, da el mismo número),
# ya que ambos representan cuántas filas se eliminarían al quedarnos con una sola copia por grupo
pct_sobrantes = dup_first / len(hoteles) * 100
print(f'Porcentaje de filas sobrantes: {pct_sobrantes:.2f}%')

# Número de grupos repetidos: la diferencia entre "todas las copias" (keep=False) y
# "las copias sobrantes" (keep='first') es exactamente el número de grupos, porque cada
# grupo de tamaño n aporta n filas a keep=False y (n-1) filas a keep='first' -- la resta
# deja 1 fila por grupo, es decir, el número de grupos.
n_grupos_repetidos = dup_false - dup_first
print(f'Número de grupos repetidos: {n_grupos_repetidos}')

# Verificación independiente con value_counts(), para confirmar el cálculo anterior
conteos_combinaciones = hoteles.value_counts(dropna=False)
n_grupos_verificacion = (conteos_combinaciones > 1).sum()
print(f'Verificación con value_counts(): {n_grupos_verificacion} grupos repetidos')
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
hoteles_copia = hoteles.copy()
# Marcamos cada fila como duplicada (sobrante) o no, usando keep='first' igual que en 1.1
hoteles_copia['es_duplicado'] = hoteles_copia.duplicated(keep='first')

# Agrupamos por hotel: sum() cuenta cuántos True (duplicados) hay, count() da el total de reservas
duplicados_por_hotel = hoteles_copia.groupby('hotel')['es_duplicado'].agg(
    filas_duplicadas='sum',
    total_reservas='count'
)
duplicados_por_hotel['pct_duplicadas'] = (
    duplicados_por_hotel['filas_duplicadas'] / duplicados_por_hotel['total_reservas'] * 100
)

duplicados_por_hotel
```

**Respuesta 1.4:**

El problema de duplicados en este dataset es grave: 31,994 de las 119,390 filas (26.80%) son copias sobrantes de otra fila idéntica en las 32 columnas, agrupadas en 8,171 grupos repetidos. Es decir, más de una de cada cuatro reservas del dataset es una copia de otra reserva que ya está registrada, lo cual es una proporción demasiado alta para ignorarla en cualquier análisis o modelo.

### 2.1 Ver los grupos repetidos (6 pts)

Filtra las filas involucradas en duplicados (`keep=False`), **ordénalas** con `sort_values()` para que las copias queden juntas y muestra las primeras 12 filas con estas columnas:

`hotel`, `lead_time`, `arrival_date_year`, `arrival_date_month`, `arrival_date_day_of_month`, `adults`, `country`, `market_segment`, `deposit_type`, `adr`, `reservation_status`.

Elige tú las columnas de ordenamiento y explica tu elección en un comentario.

```python
columnas_a_mostrar = ['hotel', 'lead_time', 'arrival_date_year', 'arrival_date_month',
                       'arrival_date_day_of_month', 'adults', 'country', 'market_segment',
                       'deposit_type', 'adr', 'reservation_status']

filas_duplicadas = hoteles[hoteles.duplicated(keep=False)].copy()

# Ordenamos por TODAS las columnas mostradas: como las filas de un mismo grupo son idénticas
# en todas ellas, ordenar por esa combinación garantiza que las copias de cada grupo queden
# juntas en el resultado (a diferencia de ordenar solo por una o dos columnas, donde grupos
# distintos con, por ejemplo, el mismo hotel y lead_time podrían intercalarse).
filas_ordenadas = filas_duplicadas.sort_values(by=columnas_a_mostrar)

filas_ordenadas[columnas_a_mostrar].head(12)
```

### 2.2 Tamaño de los grupos (6 pts)

Usa `value_counts()` sobre el `DataFrame` completo para contar cuántas veces aparece cada combinación de filas. Considera que las columnas `agent` y `company` tienen faltantes (revisa el parámetro `dropna`).

Con ese resultado:

1. Muestra **cuántos grupos** hay de cada tamaño (2 copias, 3 copias, …). Basta con los 10 tamaños más frecuentes.
2. Indica el **tamaño del grupo más grande** y muestra su combinación de valores.

```python
# dropna=False es importante: por defecto value_counts() ignora combinaciones con NaN en
# alguna columna, y aquí agent/company tienen muchos faltantes -- sin dropna=False perderíamos
# de vista justamente los grupos duplicados que comparten un valor faltante en esas columnas.
conteos_combinaciones = hoteles.value_counts(dropna=False)

# Nos quedamos solo con las combinaciones que se repiten (tamaño de grupo > 1)
grupos_repetidos = conteos_combinaciones[conteos_combinaciones > 1]

# 1. Cuántos grupos hay de cada tamaño (2 copias, 3 copias, ...) -- top 10 tamaños más frecuentes
print('Cantidad de grupos por tamaño (10 tamaños más frecuentes):')
print(grupos_repetidos.value_counts().head(10))
print()

# 2. Tamaño del grupo más grande y su combinación de valores
tamano_max = grupos_repetidos.max()
combinacion_mas_grande = grupos_repetidos.idxmax()
print(f'El grupo más grande tiene {tamano_max} copias idénticas.')
print('Su combinación de valores es:')
print(combinacion_mas_grande)
```

**Respuesta 2.4:**

Tres hallazgos lo respaldan. Primero, el grupo más grande tiene 180 copias idénticas, todas con `market_segment='Groups'`, `deposit_type='Non Refund'` y `is_canceled=1` -- es perfectamente plausible que un grupo de turistas reserve 180 habitaciones idénticas (mismo tipo de habitación, mismas fechas, mismo país, mismo canal) y que, al no existir un identificador de reserva ni de huésped, cada habitación quede registrada como una fila indistinguible de las demás. ## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# keep='first' conserva la primera aparición de cada grupo (igual criterio que usamos en 1.1-1.3);
# reset_index(drop=True) renumera el índice de 0 a n-1, sin dejar huecos por las filas eliminadas
hoteles_limpio = hoteles.drop_duplicates(keep='first').reset_index(drop=True)

filas_antes = len(hoteles)
filas_despues = len(hoteles_limpio)
filas_eliminadas = filas_antes - filas_despues

print(f'Filas antes:      {filas_antes}')
print(f'Filas después:    {filas_despues}')
print(f'Filas eliminadas: {filas_eliminadas}')
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Calculamos las medias de is_canceled y adr por hotel, y por separado el número de reservas con .size()
resumen_original = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()
resumen_original['reservas'] = hoteles.groupby('hotel').size().values

resumen_limpio = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()
resumen_limpio['reservas'] = hoteles_limpio.groupby('hotel').size().values

print('Resumen original (con duplicados):')
print(resumen_original)
print()
print('Resumen limpio (sin duplicados):')
print(resumen_limpio)
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'

# pd.concat() apila ambos resúmenes uno debajo del otro (mismo esquema de columnas)
comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True)

# Ordenamos por hotel y version para que las dos versiones de cada hotel queden juntas,
# y renumeramos el índice tras el ordenamiento
comparacion = comparacion.sort_values(by=['hotel', 'version']).reset_index(drop=True)

comparacion
```

Segundo, incluso si todos los duplicados de un grupo cayeran del mismo lado de la división, el modelo daría un peso desproporcionado a esos patrones sobrerrepresentados (como las reservas grupales canceladas) simplemente porque aparecen muchas más veces en los datos, sesgando las predicciones hacia ese tipo de reserva en detrimento de patrones menos repetidos pero igualmente válidos.

### 4.1 Verificación (5 pts)
