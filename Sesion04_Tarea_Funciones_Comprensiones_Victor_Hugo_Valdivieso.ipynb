## Datos de trabajo

Ejecuta esta celda sin modificarla — todos los ejercicios se resuelven a partir de estos datos.


````python
estaciones = [
    {"nombre": "Juárez",     "region": "Norte",  "temperatura": 31.2, "poblacion": 1512450},
    {"nombre": "Chihuahua",  "region": "Centro", "temperatura": 28.5, "poblacion": 925762},
    {"nombre": "Delicias",   "region": "Centro", "temperatura": 33.7, "poblacion": 137935},
    {"nombre": "Cuauhtémoc", "region": "Norte",  "temperatura": 22.9, "poblacion": 172387},
    {"nombre": "Parral",     "region": "Sur",    "temperatura": 26.4, "poblacion": 111200},
]

ciudades = [e["nombre"] for e in estaciones]
temperaturas = [e["temperatura"] for e in estaciones]

print(estaciones)
print(ciudades)
print(temperaturas)
````

`lecturas_semana` es una lista de listas: cada sublista trae las temperaturas de tres días para una ciudad, en el mismo orden que `ciudades` (posición 0 = Juárez, posición 1 = Chihuahua, posición 2 = Delicias).


````python
lecturas_semana = [
    [30.1, 31.5, 29.8],  # Juárez
    [27.0, 28.9, 26.5],  # Chihuahua
    [32.0, 33.7, 31.1],  # Delicias
]

print(lecturas_semana)
````

Construye la lista `reporte_ciudades`, donde cada elemento sea una cadena con el formato `"Ciudad N: nombre"`, numerando **a partir de 1** (usa el parámetro `start` de `enumerate`). Recorre `ciudades`.

## Ejercicio 2 — `zip()` + `dict()` (10 puntos)

Construye `temperatura_por_ciudad`, un diccionario que empareje cada ciudad con su temperatura, usando `dict(zip(...))` sobre `ciudades` y `temperaturas`.


````python
# ESCRIBE TU CÓDIGO AQUÍ
temperatura_por_ciudad = dict(zip(ciudades, temperaturas))

print(temperatura_por_ciudad)
````

## Ejercicio 3 — `zip()` + `enumerate()` (10 puntos)

Usando el patrón `enumerate(zip(...))`, imprime una línea por ciudad con el formato exacto:

```
1. Juárez: 31.2°C
2. Chihuahua: 28.5°C
...
```

(numeración desde 1). No necesitas guardar el resultado en una variable — solo imprime.


````python
# ESCRIBE TU CÓDIGO AQUÍ
for i, (ciudad, temperatura) in enumerate(zip(ciudades, temperaturas), start=1):
    print(f"{i}. {ciudad}: {temperatura}°C")
````

## Ejercicio 4 — `sorted()` (10 puntos)

Construye `orden_por_temperatura`: una lista de tuplas `(temperatura, ciudad)`, ordenada de **menor a mayor temperatura**.

Pista: no necesitas `key` para este ejercicio — piensa en qué compara Python primero cuando ordena una lista de tuplas, y en qué fue lo primero que le diste a `zip`.


````python
# ESCRIBE TU CÓDIGO AQUÍ
orden_por_temperatura = sorted(zip(temperaturas, ciudades))

print(orden_por_temperatura)
````

## Ejercicio 5 — `reversed()` (10 puntos)

Usando `orden_por_temperatura` del ejercicio anterior, construye `orden_descendente`: la misma lista de tuplas, pero de mayor a menor temperatura — **sin volver a ordenar**, solo invirtiendo el resultado que ya tienes.


````python
# ESCRIBE TU CÓDIGO AQUÍ
orden_descendente = list(reversed(orden_por_temperatura))

print(orden_descendente)
````

## Ejercicio 6 — Comprensión de listas (10 puntos)

Construye `ciudades_calidas`: una lista con el `nombre` de las estaciones cuya `temperatura` sea mayor a 27, usando una comprensión de listas sobre `estaciones`.


````python
# ESCRIBE TU CÓDIGO AQUÍ
ciudades_calidas = [e["nombre"] for e in estaciones if e["temperatura"] > 27]

print(ciudades_calidas)
````

## Ejercicio 7 — Comprensión de diccionarios (10 puntos)

Construye `poblacion_centro`: un diccionario `nombre: poblacion`, solo con las estaciones cuya `region` sea `"Centro"`, usando una comprensión de diccionarios con condición.


````python
# ESCRIBE TU CÓDIGO AQUÍ
poblacion_centro = {e["nombre"]: e["poblacion"] for e in estaciones if e["region"] == "Centro"}

print(poblacion_centro)
````

## Ejercicio 8 — Comprensión de conjuntos (10 puntos)

Construye `regiones_unicas`: el conjunto de todas las regiones distintas presentes en `estaciones`, usando una comprensión de conjuntos.


````python
# ESCRIBE TU CÓDIGO AQUÍ
regiones_unicas = {e["region"] for e in estaciones}

print(regiones_unicas)
print(len(regiones_unicas))
````

## Ejercicio 9 — Comprensión anidada / *flattening* (10 puntos)

Construye `temperaturas_altas`: una sola lista plana con todas las temperaturas de `lecturas_semana` (sin importar de qué ciudad vinieron) que sean mayores a 30, usando **una sola comprensión anidada** (dos `for` + un `if`, en una sola expresión).


````python
# ESCRIBE TU CÓDIGO AQUÍ
temperaturas_altas = [temp for dia in lecturas_semana for temp in dia if temp > 30]

print(temperaturas_altas)
````

# Reto integrador (10 puntos)

Construye `resumen_norte`: una lista de cadenas con el formato `"N. nombre (temperatura°C)"`, que cumpla **todas** estas condiciones:

1. Solo incluye estaciones cuya `region` sea `"Norte"`.
2. Está ordenada de mayor a menor temperatura.
3. Está numerada empezando en 1 (el número va al inicio de cada cadena, como en el Ejercicio 1).

Debes combinar al menos tres de las herramientas vistas en esta sesión (comprensión de listas, `sorted`, `enumerate`, `zip`, `reversed`). No hay una única forma de combinarlas, pero el resultado final debe ser exactamente el mismo para todos: dos elementos, con Juárez primero.


````python
# ESCRIBE TU CÓDIGO AQUÍ
norte_temp_nombre = sorted(
    [(e["temperatura"], e["nombre"]) for e in estaciones if e["region"] == "Norte"],
    reverse=True,
)
resumen_norte = [f"{i}. {nombre} ({temperatura}°C)" for i, (temperatura, nombre) in enumerate(norte_temp_nombre, start=1)]

print(resumen_norte)
````
