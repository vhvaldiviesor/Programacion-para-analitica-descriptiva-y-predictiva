```python
import pandas as pd

# Carga del dataset de taxis de Nueva York desde un repositorio público de GitHub.
# El archivo se lee con pandas; no se requiere ninguna otra biblioteca.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv'
taxis = pd.read_csv(url)

print('Dimensiones:', taxis.shape)
taxis.head()
```

## Actividad 1. Conversión y validación (20 puntos)

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
# 1.1 Conversion y validacion de pickup y dropoff

# Tipo de dato original: ambas columnas llegan como texto (object/str) desde el CSV.
print("Tipos antes de convertir:")
print(taxis[['pickup', 'dropoff']].dtypes)

# Los valores tienen el formato 'YYYY-MM-DD HH:MM:SS' (ej. 2019-03-23 20:21:09),
# asi que usamos un formato explicito en vez de dejar que pandas lo infiera.
faltantes_antes = taxis[['pickup', 'dropoff']].isna().sum()

taxis['pickup'] = pd.to_datetime(taxis['pickup'], format='%Y-%m-%d %H:%M:%S')
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format='%Y-%m-%d %H:%M:%S')

print("\nTipos despues de convertir:")
print(taxis[['pickup', 'dropoff']].dtypes)

# Si el formato no coincidiera con algun valor, pd.to_datetime con formato explicito
# lanzaria un error (en vez de inventar una fecha silenciosamente), por lo que
# comparamos nulos antes y despues como verificacion adicional.
faltantes_despues = taxis[['pickup', 'dropoff']].isna().sum()
print("\nFaltantes antes de convertir:\n", faltantes_antes)
print("\nFaltantes (NaT) despues de convertir:\n", faltantes_despues)
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# 2.1 Duracion de cada viaje

# La resta de dos columnas datetime da un Timedelta.
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

# total_seconds() da la duracion en segundos; dividimos entre 60 para minutos.
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

- Duración **negativa** (termina antes de iniciar).
- Duración **cero**.
- Duración **menor a 1 minuto** (incluye los de duración cero).
- Duración **mayor a 3 horas**.

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

```python
# 2.2 Duraciones anomalas

negativa = (taxis['duracion_min'] < 0)
cero = (taxis['duracion_min'] == 0)
menor_1min = (taxis['duracion_min'] < 1)
mayor_3h = (taxis['duracion_min'] > 180)  # 3 horas = 180 minutos

print("Duracion negativa:", negativa.sum())
print("Duracion cero:", cero.sum())
print("Duracion menor a 1 minuto:", menor_1min.sum())
print("Duracion mayor a 3 horas:", mayor_3h.sum())

print("\nViajes con duracion cero:")
display(taxis.loc[cero, ['pickup', 'dropoff', 'distance', 'fare', 'total']])
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# 2.3 Velocidad promedio

# Calculamos la velocidad para todos los renglones y luego usamos .where() para
# dejar como NaN los que tienen duracion <= 0 (donde la division no tiene sentido
# fisico o dispara division entre cero).
velocidad_bruta = taxis['distance'] / (taxis['duracion_min'] / 60)
taxis['velocidad_mph'] = velocidad_bruta.where(taxis['duracion_min'] > 0)

velocidad_alta = (taxis['velocidad_mph'] > 60)
print("Viajes con velocidad mayor a 60 mph:", velocidad_alta.sum())

top5_velocidad = taxis.sort_values('velocidad_mph', ascending=False).head(5)
print("\nTop 5 viajes por velocidad:")
display(top5_velocidad[['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']])
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# 3.1 Extraccion de componentes de fecha/hora

dias_es = {0: 'Lunes', 1: 'Martes', 2: 'Miércoles', 3: 'Jueves',
           4: 'Viernes', 5: 'Sábado', 6: 'Domingo'}

taxis['hora'] = taxis['pickup'].dt.hour
taxis['dia_num'] = taxis['pickup'].dt.dayofweek          # lunes = 0
taxis['dia_nombre'] = taxis['dia_num'].map(dias_es)
taxis['fin_semana'] = taxis['dia_num'] >= 5               # 5 = sabado, 6 = domingo
taxis['fecha'] = taxis['pickup'].dt.date

# Compara solo la parte de fecha (sin hora) de dropoff y pickup
taxis['cruza_medianoche'] = taxis['dropoff'].dt.date != taxis['pickup'].dt.date

print("Viajes que cruzan la medianoche:", taxis['cruza_medianoche'].sum())
display(taxis.loc[taxis['cruza_medianoche'], ['pickup', 'dropoff', 'duracion_min']].head())
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# 3.2 Patron por hora del dia

patron_hora = taxis.groupby('hora').agg(
    numero_viajes=('pickup', 'size'),
    mediana_duracion=('duracion_min', 'median')
)
display(patron_hora)

hora_mas_viajes = patron_hora['numero_viajes'].idxmax()
hora_menos_viajes = patron_hora['numero_viajes'].idxmin()
print(f"Hora con mas viajes: {hora_mas_viajes}:00 ({patron_hora['numero_viajes'].max()} viajes)")
print(f"Hora con menos viajes: {hora_menos_viajes}:00 ({patron_hora['numero_viajes'].min()} viajes)")

hora_mayor_dur = patron_hora['mediana_duracion'].idxmax()
hora_menor_dur = patron_hora['mediana_duracion'].idxmin()
print(f"Hora con mayor duracion mediana: {hora_mayor_dur}:00 ({patron_hora['mediana_duracion'].max():.2f} min)")
print(f"Hora con menor duracion mediana: {hora_menor_dur}:00 ({patron_hora['mediana_duracion'].min():.2f} min)")
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# 3.3 Patron por dia de la semana

patron_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    numero_viajes=('pickup', 'size'),
    mediana_duracion=('duracion_min', 'median'),
    mediana_fare=('fare', 'median'),
    fechas_distintas=('fecha', 'nunique')
).reset_index().sort_values('dia_num')  # orden lunes -> domingo

# marzo de 2019 no tiene el mismo numero de cada dia de la semana (por ejemplo,
# hay 5 viernes/sabados/domingos pero solo 4 lunes/martes/miercoles), asi que
# comparar solo el total por dia de la semana favorece a los dias con mas fechas.
patron_dia['promedio_por_fecha'] = patron_dia['numero_viajes'] / patron_dia['fechas_distintas']

display(patron_dia.set_index('dia_nombre')[
    ['numero_viajes', 'mediana_duracion', 'mediana_fare', 'fechas_distintas', 'promedio_por_fecha']
])
```

### 3.4 Interpretación (6 pts)

**Respuesta 3.4:**

Esto tiene sentido: a las 18:00 hay muchísimos viajes cortos dentro de zonas concurridas (salida de oficinas, viajes de conexión), mientras que al mediodía puede haber una mezcla de viajes de mayor distancia (comidas de negocio, trámites) combinada con el tráfico típico del mediodía, que alarga la duración incluso si el número de viajes no es el más alto del día.

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.
